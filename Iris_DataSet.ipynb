```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(data=df, x="SepalLengthCm", bins=30, kde=True)

# Gemiddelde (Mean) line
plt.axvline(
    df["SepalLengthCm"].mean(),
    color="red",
    linestyle="--",
    label=f"Mean ({df['SepalLengthCm'].mean():.2f})"
)

# Mediaan (Median) line
plt.axvline(
    df["SepalLengthCm"].median(),
    color="green",
    linestyle="--",
    label=f"Median ({df['SepalLengthCm'].median():.2f})"
)

# Modus (Mode) line
plt.axvline(
    df["SepalLengthCm"].mode()[0],
    color="blue",
    linestyle="--",
    label=f"Mode ({df['SepalLengthCm'].mode()[0]:.2f})"
)

plt.title("Sepal Uzunluğunun Paylanması (Sepal Length Distribution)")
plt.xlabel("Sepal Uzunluğu (cm)")
plt.ylabel("Tezlik (Sayı)")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(data=df, x="PetalLengthCm", bins=30, kde=True)

# Gemiddelde (Mean) line
plt.axvline(
    df["PetalLengthCm"].mean(),
    color="red",
    linestyle="--",
    label=f"Mean ({df['PetalLengthCm'].mean():.2f})"
)

# Mediaan (Median) line
plt.axvline(
    df["PetalLengthCm"].median(),
    color="green",
    linestyle="--",
    label=f"Median ({df['PetalLengthCm'].median():.2f})"
)

# Modus (Mode) line
plt.axvline(
    df["PetalLengthCm"].mode()[0],
    color="blue",
    linestyle="--",
    label=f"Mode ({df['PetalLengthCm'].mode()[0]:.2f})"
)

plt.title("Petal Uzunluğunun Paylanması (Length Distribution)")
plt.xlabel("Petal Uzunluğu (cm)")
plt.ylabel("Tezlik (Sayı)")
plt.legend()
plt.show()
```

```python
#Kvartil və İQR tapılması
q1=df["SepalLengthCm"].quantile(0.25)
q2=df["SepalLengthCm"].quantile(0.50)
q3=df["SepalLengthCm"].quantile(0.75)

iqr=q3-q1
```

```python
print(q1)
print(q2)
print(q3)
print(iqr)
```

```python
#Yuxarı və aşağı sərhədlərin təyin olunması
lower_bound=q1-iqr*1.5
upper_bound=q3+iqr*1.5
print(f'Lower_bound :{lower_bound}')
print(f'Upper_bound :{upper_bound}')
```

```python
outliers = df[(df["SepalLengthCm"] < lower_bound) | (df["SepalLengthCm"] > upper_bound)]
```

```python
#Kvartillərin tapılması
q1 = df["SepalLengthCm"].quantile(0.25)
q2 = df["SepalLengthCm"].quantile(0.50)  # Mediana
q3 = df["SepalLengthCm"].quantile(0.75)

#IQR hesablanması
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Lower_bound: {lower_bound:.2f}")
print(f"Upper_bound: {upper_bound:.2f}")

#Outliers
outliers = df[(df["SepalLengthCm"] < lower_bound) | (df["SepalLengthCm"] > upper_bound)]
print(f"Kənar qiymət sayı: {len(outliers)}")
```
